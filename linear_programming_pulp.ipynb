# 2. $n$-dimensional linear program in real numbers

Arcelor-Mittal, one of the world’s leading steel manufacturer, has opened a new plant in Greece. It manufactures 9 alloys made from 3 raw materials&nbsp;: lead, zinc and tin.
Here is an array containing the percentages of each material in the alloy and its cost per unit&nbsp;:

| Alloy | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 | 9 |
|---------------|-----|-----|-----|-----|-----|-----|-----|-----|-----|
|**% lead** | 20 | 50 | 30 | 30 | 30 | 60 | 40 | 10 | 10 |
|**% zinc** | 30 | 40 | 20 | 40 | 30 | 30 | 50 | 30 | 10 |
|**% tin** | 50 | 10 | 50 | 30 | 40 | 10 | 10 | 60 | 80 |
|**Unit cost** | 7.3 | 6.9 | 7.3 | 7.5 | 7.6 | 6.0 | 5.8 | 4.3 | 4.1 |

The R&D department is looking for a ‘new alloy’ at a minimum price, which would be a mixture of 9 other alloys, with the constraints of having a maximum of 30 % lead, 30 % zinc and 40 % tin.

## 2.2 Problem-solving
We are dealing with a 9-variable problem. Since the graphic problem-solving method cannot be applied here, we will opt for the [PuLP](https://pythonhosted.org/PuLP/) solver, which will automatically calculate the solution by exploiting the property that we have identified in the previous exercise.


We will draw inspiration from [a basic example](https://www.coin-or.org/PuLP/CaseStudies/a_blending_problem.html) to implement and solve the model&nbsp;:



````python
! pip install pulp
from pulp import *
import numpy as np

# Variables of the problem (fractions of each alloy)
X1 = LpVariable("X_1", 0, None)
X2 = LpVariable("X_2", 0, None)
X3 = LpVariable("X_3", 0, None)
X4 = LpVariable("X_4", 0, None)
X5 = LpVariable("X_5", 0, None)
X6 = LpVariable("X_6", 0, None)
X7 = LpVariable("X_7", 0, None)
X8 = LpVariable("X_8", 0, None)
X9 = LpVariable("X_9", 0, None)

# Problem
prob = LpProblem("Alloy_Mixture_MinCost", LpMinimize)

# Objective function (minimize cost)
prob += (7.3*X1 + 6.9*X2 + 7.3*X3 + 7.5*X4 + 7.6*X5 +
         6.0*X6 + 5.8*X7 + 4.3*X8 + 4.1*X9), "Total_Cost"

# Constraints
# Maximum lead ≤ 30%
prob += (0.20*X1 + 0.50*X2 + 0.30*X3 + 0.30*X4 + 0.30*X5 +
         0.60*X6 + 0.40*X7 + 0.10*X8 + 0.10*X9 <= 0.30), "Lead_Max"

# Maximum zinc ≤ 30%
prob += (0.30*X1 + 0.40*X2 + 0.20*X3 + 0.40*X4 + 0.30*X5 +
         0.30*X6 + 0.50*X7 + 0.30*X8 + 0.10*X9 <= 0.30), "Zinc_Max"

# Maximum tin ≤ 40%
prob += (0.50*X1 + 0.10*X2 + 0.50*X3 + 0.30*X4 + 0.40*X5 +
         0.10*X6 + 0.10*X7 + 0.60*X8 + 0.80*X9 <= 0.40), "Tin_Max"

# Sum of fractions = 1 (optional if mixture is in fractions)
prob += X1 + X2 + X3 + X4 + X5 + X6 + X7 + X8 + X9 == 1, "Sum_to_1"

# Solve
prob.solve()

# Print status and minimum cost
print("Status:", LpStatus[prob.status])
print("Minimum cost =", value(prob.objective))

# Print variable results
for v in prob.variables():
    print("%s = %.4f" % (v.name, v.varValue), end=', ')
````

## 3.2 Problem-solving

Since we have modelled the problem in linear form, why not test the Simplex algorithm on it&nbsp;? Let’s take the data from our student’s Smartphone. Beware of the bounds of our variables&nbsp;!

````python
from pulp import *

# Variables of the problem (binary: 0 or 1)
X1 = LpVariable("X_1", 0, 1)
X2 = LpVariable("X_2", 0, 1)
X3 = LpVariable("X_3", 0, 1)
X4 = LpVariable("X_4", 0, 1)
X5 = LpVariable("X_5", 0, 1)
X6 = LpVariable("X_6", 0, 1)
X7 = LpVariable("X_7", 0, 1)
X8 = LpVariable("X_8", 0, 1)
X9 = LpVariable("X_9", 0, 1)
X10 = LpVariable("X_10", 0, 1)

# Problem
prob = LpProblem("Music_Playlist_Knapsack", LpMaximize)

# Objective: maximize total rating
prob += (2*X1 + 5*X2 + 1*X3 + 2*X4 + 3*X5 + 5*X6 + 2*X7 + 3*X8 + 1*X9 + 1*X10), "Total_Rating"

# Constraint: total size <= 500 MB
prob += (40*X1 + 52*X2 + 27*X3 + 68*X4 + 80*X5 + 62*X6 + 63*X7 + 53*X8 + 56*X9 + 59*X10 <= 500), "Storage_Limit"

# Solve
prob.solve()

# Print status and total rating
print(LpStatus[prob.status])
print("Valeur totale =", value(prob.objective))

# Print variables and total weight
poids_total = 0
weights = [40, 52, 27, 68, 80, 62, 63, 53, 56, 59]
for idx, v in enumerate(prob.variables()):
    print("X%d=%d" % (idx+1, v.varValue), end=', ')
    poids_total += v.varValue * weights[idx]
print("\npoids = %d" % poids_total)
````
